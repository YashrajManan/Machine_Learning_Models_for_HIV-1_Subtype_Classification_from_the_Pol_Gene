# Ampmcnemar — Amplicon McNemar: SVM vs every CNN, intact and under dropout

**Why this notebook exists.** Section 3.8/4.1's two amplicon claims -- "all five models are
statistically indistinguishable on the intact PR+RT amplicon" and "the linear SVM is by a wide,
seed-consistent margin the most robust to a simulated 200 bp dropout" -- are both currently
supported only by comparing means and standard deviations across 3 seeds, not by a paired test on
matched per-sequence predictions. The paper's Limitations section flags this explicitly as unclosed.
**This notebook closes it**, on the CD-HIT-97% grouping this paper treats as primary everywhere else.
It reruns all five models (linear SVM, random forest, ResNet-18, VGG-16, Inception-v1) on two
amplicon inputs -- the intact PR+RT amplicon and the same amplicon with a simulated 200 bp internal
dropout -- over 3 seeds, saves every per-sequence prediction, and runs an exact McNemar test (SVM vs
each CNN, both conditions, Holm-Bonferroni corrected) right here in the notebook so you get the
p-values without a second round trip.

**Needs:** `hxb2_pr_rt_coords.csv` from AmpliconPrimer (the real PR-RT amplicon extraction notebook),
attached as a Kaggle Dataset. Without it, this notebook cannot run at all -- unlike Part B below
(PR-alone/RT-alone), this notebook's whole purpose is the combined amplicon.

**GPU needed:** yes, for the three CNNs, but the amplicon test set is small (about 4,500-4,700
sequences depending on split/seed, versus ~24,000 for the main fragmentation sweep), so this is
much cheaper than Fragsweep/Nineseed. 3 seeds x 2 amplicon inputs x 3 CNNs = 18 CNN fits, at the same
per-fit cost as Fragsweep/Nineseed (~2-3 min each on a T4 GPU) -- **roughly 40-60 minutes of GPU time**,
comfortably inside whatever is left of this week's quota.

**Dataset.** All notebooks below read the *same* dataset you already used for the main experiments and the 5-model fragment sweep:

`/kaggle/input/datasets/yashrajrajarishi/subtype-classification/prepared_sequences_clean.csv`

with columns `sequence`, `subtype`, `cluster` (the MinHash cluster id). Attach that Kaggle Dataset to the notebook the same way you did for the fragment-test notebook (Add Data -> Your Datasets -> subtype-classification). **No new download is needed** for this file. Where a test needs something extra (a reference genome, an external tool), that is called out explicitly at the top of the notebook, with the exact cell that fetches or installs it -- it downloads automatically inside the notebook, you do not need to upload anything yourself.

### 2. Load the full dataset and run CD-HIT-97% (same command/args as Fragsweep/Nineseed, so clusters match)

```python

DATA_PATH = "/workspace/data/prepared_sequences_clean.csv"
df = pd.read_csv(DATA_PATH).reset_index(drop=True)
X_all = df["sequence"].values
n = len(df)

with open(f"{OUT}/pol_ampmcnemar.fasta", "w") as f:
    for i, s in enumerate(X_all):
        f.write(f">{i}\n{s}\n")

def parse_cdhit_clusters(clstr_path, n_seqs):
    labels = np.full(n_seqs, -1, dtype=int)
    cid = -1
    with open(clstr_path) as f:
        for line in f:
            if line.startswith(">Cluster"):
                cid += 1
            else:
                idx = int(line.split(">")[1].split("...")[0])
                labels[idx] = cid
    assert (labels >= 0).all()
    return labels

out = f"{OUT}/pol_ampmcnemar_cdhit_97"
cmd = f"cd-hit-est -i {OUT}/pol_ampmcnemar.fasta -o {out} -c 0.97 -n 10 -M 16000 -T 4 -d 0"
r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
assert r.returncode == 0 and os.path.exists(out + ".clstr"), f"CD-HIT failed: {r.returncode}"
groups_all = parse_cdhit_clusters(out + ".clstr", n)
print(f"CD-HIT 97% identity: {len(set(groups_all))} clusters (should match Fragsweep/Nineseed's 7,182)")
```

### 3. Load the AmpliconPrimer PR/RT coordinates and build the two amplicon inputs

```python

COORDS_PATH = "/workspace/data/hxb2_pr_rt_coords.csv"
assert os.path.exists(COORDS_PATH), (
    f"{COORDS_PATH} not found. This notebook needs the AmpliconPrimer output (hxb2_pr_rt_coords.csv) attached "
    "as a Kaggle Dataset -- there is no fallback for this one, unlike Part B below.")

coords_df = pd.read_csv(COORDS_PATH)
good = coords_df.dropna(subset=["pr", "rt"]).reset_index(drop=True)
print(f"{len(good)} sequences with both PR and RT extracted")

def dropout_200nt(seq, rng):
    if len(seq) <= 220:
        return seq
    start = rng.integers(50, len(seq) - 200 - 50)
    return seq[:start] + seq[start+200:]

rng0 = np.random.default_rng(0)
good["pr_rt_combined"] = good["pr"] + good["rt"]
good["pr_rt_dropout"] = good["pr_rt_combined"].apply(lambda s: dropout_200nt(s, rng0))

le = LabelEncoder(); y_good = le.fit_transform(good["subtype"].values)
n_cls = len(le.classes_)
INPUTS = ["pr_rt_combined", "pr_rt_dropout"]

# Map amplicon rows back to the CD-HIT-97% cluster ids computed on the full dataset above,
# so this test uses the same primary grouping as the rest of the paper (an earlier amplicon test used the MinHash grouping instead).
# Tries several strategies in order of reliability and VERIFIES each before trusting it --
# never silently falls back to the wrong split without saying so.
def build_amplicon_groups(coords_df, good, df, X_all, groups_all):
    n_full = len(df)
    for col in ["orig_index", "orig_idx", "index", "row_index", "source_index", "full_index"]:
        if col in good.columns:
            try:
                idx = good[col].astype(int).values
                if idx.min() >= 0 and idx.max() < n_full:
                    return groups_all[idx], f"used explicit '{col}' column"
            except Exception:
                pass
    for col in ["sequence", "full_sequence", "seq", "pol_sequence", "orig_sequence"]:
        if col in good.columns:
            idx_map = {s: i for i, s in enumerate(X_all)}
            mapped = good[col].map(idx_map)
            if mapped.notna().all():
                return groups_all[mapped.astype(int).values], f"matched on '{col}' full-sequence text"
    if len(coords_df) == n_full and len(good) == n_full:
        if "subtype" in good.columns and "subtype" in df.columns:
            match_rate = (good["subtype"].values == df["subtype"].values).mean()
            if match_rate > 0.99:
                return groups_all[good.index.values], f"positional alignment, verified via subtype match ({match_rate:.1%})"
            else:
                print(f"Positional alignment REJECTED: subtype columns only agree {match_rate:.1%} row-by-row.")
        else:
            print("Same row count as the full dataset, but no 'subtype' column in one or both files to "
                  "verify row order -- not risking an unverified positional mapping.")
    print(f"Could not verify a mapping back to CD-HIT-97% clusters. Columns available in "
          f"hxb2_pr_rt_coords.csv: {list(coords_df.columns)}. Send these column names back and the "
          "mapping can be fixed precisely instead of guessed at.")
    return None, "no verified mapping"

amplicon_groups, mapping_method = build_amplicon_groups(coords_df, good, df, X_all, groups_all)
assert amplicon_groups is not None, (
    "Stopping here -- see the message above. This notebook's whole purpose is a CD-HIT-97% amplicon "
    "test, so it refuses to silently fall back to MinHash the way an earlier run of this notebook did.")
print(f"Mapping method used: {mapping_method}")
```

### 4. Split helper (with validation carve-out for the CNNs) and shared TF-IDF encoder

```python

MAX_TFIDF_FEATS = 512
SEEDS = [0, 1, 2]
EPOCHS = 120     # matches the original per-architecture notebooks (was 60 -- bug, now fixed)
BATCH  = 48

def split_grouped_with_val(X, y, grp, seed):
    s1 = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=seed)
    trv, te = next(s1.split(X, y, grp))
    s2 = StratifiedGroupKFold(n_splits=4, shuffle=True, random_state=seed)
    tr_rel, va_rel = next(s2.split(X[trv], y[trv], grp[trv]))
    return trv[tr_rel], trv[va_rel], te

def fit_tfidf(seq_tr, seq_va, seq_te):
    vec = TfidfVectorizer(analyzer="char", ngram_range=(6,6), max_features=MAX_TFIDF_FEATS, lowercase=False)
    Xtr = vec.fit_transform(seq_tr); Xva = vec.transform(seq_va); Xte = vec.transform(seq_te)
    return Xtr, Xva, Xte
```

### 5. SVM + RF, both amplicon inputs, 3 seeds -- CPU only

```python

rows = []
pred_rows = []
splits_cache = {}

for col in INPUTS:
    for seed in SEEDS:
        tr, va, te = split_grouped_with_val(good.index.values, y_good, amplicon_groups, seed)
        splits_cache[(col, seed)] = (tr, va, te)
        Xtr, Xva, Xte = fit_tfidf(good.iloc[tr][col].values, good.iloc[va][col].values, good.iloc[te][col].values)

        Xtrva = np.vstack([Xtr.toarray(), Xva.toarray()])
        ytrva = np.concatenate([y_good[tr], y_good[va]])
        yte = y_good[te]
        for model_name, clf in [
            ("svm", LinearSVC(class_weight="balanced", random_state=seed, max_iter=20000)),
            ("rf", RandomForestClassifier(n_estimators=400, class_weight="balanced", random_state=seed, n_jobs=-1)),
        ]:
            clf.fit(Xtrva, ytrva)
            pred = clf.predict(Xte)
            rows.append(dict(input=col, seed=seed, model=model_name,
                              f1=f1_score(yte, pred, average="macro"),
                              bal_acc=balanced_accuracy_score(yte, pred)))
            pred_rows.extend([dict(input=col, seed=seed, model=model_name,
                                    test_idx=int(good.index.values[i]), y_true=int(t), y_pred=int(p))
                               for i, t, p in zip(te, yte, pred)])
    pd.DataFrame(rows).to_csv(f"{OUT}/ampmcnemar_classical_raw.csv", index=False)
    pd.DataFrame(pred_rows).to_csv(f"{OUT}/ampmcnemar_classical_predictions.csv", index=False)
    print(f"[classical] input={col} done")
```

### 6. CNNs, both amplicon inputs, 3 seeds -- needs GPU. Same architectures/training loop as Fragsweep/Nineseed.

```python


class StochasticDepth(Layer):
    def __init__(self, p=0.85, **kw): super().__init__(**kw); self.p = p
    def call(self, x, training=None):
        if not training: return x
        r = tf.random.uniform([tf.shape(x)[0], 1, 1])
        return x * tf.floor(self.p + r) / self.p
    def get_config(self): return {**super().get_config(), "p": self.p}

def res_block(x, f, stride=1, sp=0.85, dr=0.1):
    sc = x
    x = Conv1D(f, 3, strides=stride, padding="same", use_bias=False)(x)
    x = ReLU()(BatchNormalization()(x)); x = SpatialDropout1D(dr)(x)
    x = BatchNormalization()(Conv1D(f, 3, padding="same", use_bias=False)(x))
    if stride != 1 or sc.shape[-1] != f:
        sc = BatchNormalization()(Conv1D(f, 1, strides=stride, padding="same", use_bias=False)(sc))
    return ReLU()(Add()([StochasticDepth(sp)(x), sc]))

def build_resnet18(input_len, n_classes):
    inp = Input((input_len, 1)); x = inp
    x = ReLU()(BatchNormalization()(Conv1D(32, 7, strides=2, padding="same", use_bias=False)(x)))
    x = MaxPooling1D(3, strides=2, padding="same")(x)
    for f, sp in [(32, .95), (32, .92)]:  x = res_block(x, f, 1, sp)
    for f, sp in [(64, .88), (64, .85)]:  x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    for f, sp in [(128, .80), (128, .78)]: x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    for f, sp in [(256, .75), (256, .72)]: x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

def build_vgg16(input_len, n_classes):
    def conv_block(x, f, n_convs):
        for _ in range(n_convs):
            x = ReLU()(BatchNormalization()(Conv1D(f, 3, padding="same", use_bias=False)(x)))
        return MaxPooling1D(2, padding="same")(x)
    inp = Input((input_len, 1))
    x = conv_block(inp, 32, 2)
    x = conv_block(x,   64, 2)
    x = conv_block(x,  128, 3)
    x = conv_block(x,  256, 3)
    x = conv_block(x,  256, 3)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    x = Dropout(0.4)(ReLU()(Dense(256)(x)))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

def inception_module(x, f1, f3r, f3, f5r, f5, fp):
    b1 = Conv1D(f1, 1, padding="same", activation="relu")(x)
    b3 = Conv1D(f3, 3, padding="same", activation="relu")(Conv1D(f3r, 1, padding="same", activation="relu")(x))
    b5 = Conv1D(f5, 5, padding="same", activation="relu")(Conv1D(f5r, 1, padding="same", activation="relu")(x))
    bp = Conv1D(fp, 1, padding="same", activation="relu")(MaxPooling1D(3, strides=1, padding="same")(x))
    return Concatenate()([b1, b3, b5, bp])

def build_inception(input_len, n_classes):
    inp = Input((input_len, 1))
    x = MaxPooling1D(3, strides=2, padding="same")(
            ReLU()(BatchNormalization()(Conv1D(32, 7, strides=2, padding="same")(inp))))
    x = inception_module(x, 16, 24, 32, 4,  8,  8)
    x = inception_module(x, 32, 32, 48, 8, 24, 16)
    x = MaxPooling1D(3, strides=2, padding="same")(x)
    x = inception_module(x, 48, 24, 52, 8, 24, 16)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

BUILDERS = {"resnet18": build_resnet18, "vgg16": build_vgg16, "inception": build_inception}
DEEP_MODELS = ["resnet18", "vgg16", "inception"]

def run_deep_original(model_name, Xtr, ytr, Xva, yva, Xte, yte, seed, n_classes):
    """Exact training recipe from the original per-architecture notebooks (verified byte-identical
    in Phase A, same code already used in Fragsweep/Ambig/Enc): per-feature StandardScaler (fit on train
    only), one-hot labels, CategoricalCrossentropy(label_smoothing=0.1), Adam(3e-4), up to 120
    epochs, EarlyStopping(patience=15) + ReduceLROnPlateau(factor=0.5, patience=6, min_lr=1e-6)."""
    tf.random.set_seed(seed)
    sc = StandardScaler()
    Xtr_d = sc.fit_transform(np.asarray(Xtr.todense() if hasattr(Xtr, "todense") else Xtr))[..., None]
    Xva_d = sc.transform(np.asarray(Xva.todense() if hasattr(Xva, "todense") else Xva))[..., None]
    Xte_d = sc.transform(np.asarray(Xte.todense() if hasattr(Xte, "todense") else Xte))[..., None]
    cw = dict(enumerate(compute_class_weight("balanced", classes=np.unique(ytr), y=ytr)))
    model = BUILDERS[model_name](Xtr_d.shape[1], n_classes)
    model.fit(Xtr_d, to_categorical(ytr, n_classes),
              validation_data=(Xva_d, to_categorical(yva, n_classes)),
              epochs=EPOCHS, batch_size=BATCH, class_weight=cw, shuffle=True, verbose=0,
              callbacks=[EarlyStopping("val_loss", patience=15, restore_best_weights=True),
                         ReduceLROnPlateau("val_loss", factor=0.5, patience=6, min_lr=1e-6)])
    prob = model.predict(Xte_d, verbose=0)
    pred = prob.argmax(1)
    f1 = f1_score(yte, pred, average="macro"); bal = balanced_accuracy_score(yte, pred)
    tf.keras.backend.clear_session()
    return f1, bal, pred

DEEP_RAW_PATH = f"{OUT}/ampmcnemar_deep_raw.csv"
if os.path.exists(DEEP_RAW_PATH):
    _existing_deep = pd.read_csv(DEEP_RAW_PATH)
    deep_done_keys = set(zip(_existing_deep["input"], _existing_deep["seed"], _existing_deep["model"]))
    deep_rows = _existing_deep.to_dict("records")
    print(f"Resuming deep models: {len(deep_rows)} rows already done.")
else:
    deep_rows = []
    deep_done_keys = set()

deep_pred_rows = []
for (col, seed), (tr, va, te) in splits_cache.items():
    remaining = [m for m in DEEP_MODELS if (col, seed, m) not in deep_done_keys]
    if not remaining:
        continue
    Xtr, Xva, Xte = fit_tfidf(good.iloc[tr][col].values, good.iloc[va][col].values, good.iloc[te][col].values)
    ytr, yva, yte = y_good[tr], y_good[va], y_good[te]
    for m in remaining:
        f1, bal, pred = run_deep_original(m, Xtr, ytr, Xva, yva, Xte, yte, seed, n_cls)
        deep_rows.append(dict(input=col, seed=seed, model=m, f1=f1, bal_acc=bal))
        deep_pred_rows.extend([dict(input=col, seed=seed, model=m,
                                     test_idx=int(good.index.values[i]), y_true=int(t), y_pred=int(p))
                                for i, t, p in zip(te, yte, pred)])
        pd.DataFrame(deep_rows).to_csv(DEEP_RAW_PATH, index=False)   # incremental save
        print(f"[deep] input={col} seed={seed} model={m} f1={f1:.4f} done")
    pd.DataFrame(deep_pred_rows).to_csv(f"{OUT}/ampmcnemar_deep_predictions.csv", index=False)

ampmcnemar_all = pd.concat([pd.DataFrame(rows), pd.DataFrame(deep_rows)], ignore_index=True)
ampmcnemar_all_predictions = pd.concat([pd.DataFrame(pred_rows), pd.DataFrame(deep_pred_rows)], ignore_index=True)
ampmcnemar_all.to_csv(f"{OUT}/ampmcnemar_all_raw.csv", index=False)
ampmcnemar_all_predictions.to_csv(f"{OUT}/ampmcnemar_all_predictions.csv", index=False)
print(ampmcnemar_all.groupby(["input","model"])[["f1","bal_acc"]].agg(["mean","std"]).round(4))
```

### 7. McNemar test, SVM vs each CNN, both amplicon inputs, per seed, Holm-Bonferroni corrected

Answers both flagged claims directly: are the models really tied on the intact amplicon (`pr_rt_combined`), and is the SVM really more robust under dropout (`pr_rt_dropout`)?

```python

try:
    from statsmodels.stats.contingency_tables import mcnemar
    from statsmodels.stats.multitest import multipletests
except ImportError:
    import subprocess as sp
    sp.run(["pip", "install", "statsmodels", "-q"])
    from statsmodels.stats.contingency_tables import mcnemar
    from statsmodels.stats.multitest import multipletests

def mcnemar_pair(a_model, b_model, inp, seed):
    A = ampmcnemar_all_predictions[(ampmcnemar_all_predictions.model==a_model)&(ampmcnemar_all_predictions.input==inp)&(ampmcnemar_all_predictions.seed==seed)].set_index("test_idx")
    B = ampmcnemar_all_predictions[(ampmcnemar_all_predictions.model==b_model)&(ampmcnemar_all_predictions.input==inp)&(ampmcnemar_all_predictions.seed==seed)].set_index("test_idx")
    common = A.index.intersection(B.index)
    a_correct = (A.loc[common].y_true == A.loc[common].y_pred)
    b_correct = (B.loc[common].y_true == B.loc[common].y_pred)
    n10 = int(((a_correct)&(~b_correct)).sum())
    n01 = int(((~a_correct)&(b_correct)).sum())
    res = mcnemar([[0, n10],[n01, 0]], exact=True)
    return len(common), n10, n01, res.pvalue

results = []
for inp in INPUTS:
    for cnn in DEEP_MODELS:
        for seed in SEEDS:
            n, n10, n01, p = mcnemar_pair("svm", cnn, inp, seed)
            results.append(dict(input=inp, cnn=cnn, seed=seed, n=n,
                                 svm_only_correct=n10, cnn_only_correct=n01, p_raw=p))

res_df = pd.DataFrame(results)
res_df["p_holm"] = multipletests(res_df["p_raw"], method="holm")[1]
res_df["significant"] = res_df["p_holm"] < 0.05
res_df.to_csv(f"{OUT}/ampmcnemar_mcnemar_results.csv", index=False)
pd.set_option("display.max_rows", None)
print(res_df.round(5))
print()
print("Summary -- fraction of (cnn, seed) cells significant, by input:")
print(res_df.groupby("input")["significant"].mean())
```

# Pralone — PR-alone and RT-alone on CD-HIT-97%, closing Table 11's mixed-run problem

**Why this notebook exists.** Table 11 reports four rows: PR alone, RT alone, PR+RT combined, and
PR+RT under dropout. The PR+RT-combined and dropout rows were regenerated on the CD-HIT-97% split
(Part A / Ampmcnemar above) after a coordinate-mapping bug was fixed that had silently fallen back to
the older MinHash grouping. **The PR-alone and RT-alone rows have not been regenerated on this
split** -- they still reflect an earlier run on the MinHash split, which is why their standard
deviations (0.004-0.006) are noticeably tighter than the two regenerated rows (0.022-0.029) for the
same models in the same table. This notebook closes that gap by rerunning PR-alone and RT-alone on
the same CD-HIT-97% split and verified coordinate mapping as Part A, so all four Table 11 rows are
finally on a consistent footing.


### 1. Imports and install CD-HIT

```python

import warnings; warnings.filterwarnings("ignore")
import os, subprocess, shutil
import numpy as np, pandas as pd

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import f1_score, balanced_accuracy_score

OUT = "/workspace/outputs/pralone"
os.makedirs(OUT, exist_ok=True)

if shutil.which("cd-hit-est") is None:
    print("cd-hit-est not found, installing...")
    subprocess.run("conda install -y -c bioconda cd-hit 2>/dev/null || true", shell=True)
    if shutil.which("cd-hit-est") is None:
        subprocess.run(
            "wget -q https://github.com/weizhongli/cdhit/releases/download/V4.8.1/cd-hit-v4.8.1-2019-0228.tar.gz "
            "&& tar xzf cd-hit-v4.8.1-2019-0228.tar.gz && cd cd-hit-v4.8.1-2019-0228 && make -j4 "
            "&& cp cd-hit-est /usr/local/bin/ && cp cd-hit /usr/local/bin/", shell=True)
print("cd-hit-est available:", shutil.which("cd-hit-est") is not None)

import tensorflow as tf
from tensorflow.keras.layers import (Input, Conv1D, BatchNormalization, ReLU, Add,
    GlobalAveragePooling1D, Dropout, Dense, MaxPooling1D,
    SpatialDropout1D, Layer, Concatenate, MaxPool1D)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import CategoricalCrossentropy
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.utils import to_categorical

```

### 2. Load the full dataset and run CD-HIT-97% (same command/args as Fragsweep/Nineseed/Ampmcnemar, so clusters match)

```python

DATA_PATH = "/workspace/data/prepared_sequences_clean.csv"
df = pd.read_csv(DATA_PATH).reset_index(drop=True)
X_all = df["sequence"].values
n = len(df)

with open(f"{OUT}/pol_pralone.fasta", "w") as f:
    for i, s in enumerate(X_all):
        f.write(f">{i}\n{s}\n")

def parse_cdhit_clusters(clstr_path, n_seqs):
    labels = np.full(n_seqs, -1, dtype=int)
    cid = -1
    with open(clstr_path) as f:
        for line in f:
            if line.startswith(">Cluster"):
                cid += 1
            else:
                idx = int(line.split(">")[1].split("...")[0])
                labels[idx] = cid
    assert (labels >= 0).all()
    return labels

out = f"{OUT}/pol_pralone_cdhit_97"
cmd = f"cd-hit-est -i {OUT}/pol_pralone.fasta -o {out} -c 0.97 -n 10 -M 16000 -T 4 -d 0"
r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
assert r.returncode == 0 and os.path.exists(out + ".clstr"), f"CD-HIT failed: {r.returncode}"
groups_all = parse_cdhit_clusters(out + ".clstr", n)
print(f"CD-HIT 97% identity: {len(set(groups_all))} clusters (should match Fragsweep/Nineseed/Ampmcnemar's 7,182)")
```

### 3. Load the AmpliconPrimer PR/RT coordinates. Inputs this time: PR alone and RT alone (not combined).

```python

COORDS_PATH = "/workspace/data/hxb2_pr_rt_coords.csv"
assert os.path.exists(COORDS_PATH), (
    f"{COORDS_PATH} not found. This notebook needs the AmpliconPrimer output (hxb2_pr_rt_coords.csv) attached "
    "as a Kaggle Dataset.")

coords_df = pd.read_csv(COORDS_PATH)
good = coords_df.dropna(subset=["pr", "rt"]).reset_index(drop=True)
print(f"{len(good)} sequences with both PR and RT extracted")

le = LabelEncoder(); y_good = le.fit_transform(good["subtype"].values)
n_cls = len(le.classes_)
INPUTS = ["pr", "rt"]   # <-- the two inputs Table 11 still needs on this split

# Identical verified mapper to Ampmcnemar above -- tries several strategies and refuses to guess if none verify.
def build_amplicon_groups(coords_df, good, df, X_all, groups_all):
    n_full = len(df)
    for col in ["orig_index", "orig_idx", "index", "row_index", "source_index", "full_index"]:
        if col in good.columns:
            try:
                idx = good[col].astype(int).values
                if idx.min() >= 0 and idx.max() < n_full:
                    return groups_all[idx], f"used explicit '{col}' column"
            except Exception:
                pass
    for col in ["sequence", "full_sequence", "seq", "pol_sequence", "orig_sequence"]:
        if col in good.columns:
            idx_map = {s: i for i, s in enumerate(X_all)}
            mapped = good[col].map(idx_map)
            if mapped.notna().all():
                return groups_all[mapped.astype(int).values], f"matched on '{col}' full-sequence text"
    if len(coords_df) == n_full and len(good) == n_full:
        if "subtype" in good.columns and "subtype" in df.columns:
            match_rate = (good["subtype"].values == df["subtype"].values).mean()
            if match_rate > 0.99:
                return groups_all[good.index.values], f"positional alignment, verified via subtype match ({match_rate:.1%})"
            else:
                print(f"Positional alignment REJECTED: subtype columns only agree {match_rate:.1%} row-by-row.")
        else:
            print("Same row count as the full dataset, but no 'subtype' column in one or both files to "
                  "verify row order -- not risking an unverified positional mapping.")
    print(f"Could not verify a mapping back to CD-HIT-97% clusters. Columns available in "
          f"hxb2_pr_rt_coords.csv: {list(coords_df.columns)}. Send these column names back and the "
          "mapping can be fixed precisely instead of guessed at.")
    return None, "no verified mapping"

amplicon_groups, mapping_method = build_amplicon_groups(coords_df, good, df, X_all, groups_all)
assert amplicon_groups is not None, "Stopping here -- see the message above."
print(f"Mapping method used: {mapping_method}")
```

### 4. Split helper (with validation carve-out for the CNNs) and shared TF-IDF encoder

```python

MAX_TFIDF_FEATS = 512
SEEDS = [0, 1, 2]
EPOCHS = 120     # matches the original per-architecture notebooks (was 60 -- bug, now fixed)
BATCH  = 48

def split_grouped_with_val(X, y, grp, seed):
    s1 = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=seed)
    trv, te = next(s1.split(X, y, grp))
    s2 = StratifiedGroupKFold(n_splits=4, shuffle=True, random_state=seed)
    tr_rel, va_rel = next(s2.split(X[trv], y[trv], grp[trv]))
    return trv[tr_rel], trv[va_rel], te

def fit_tfidf(seq_tr, seq_va, seq_te):
    vec = TfidfVectorizer(analyzer="char", ngram_range=(6,6), max_features=MAX_TFIDF_FEATS, lowercase=False)
    Xtr = vec.fit_transform(seq_tr); Xva = vec.transform(seq_va); Xte = vec.transform(seq_te)
    return Xtr, Xva, Xte
```

### 5. SVM + RF, PR alone and RT alone, 3 seeds -- CPU only

```python

rows = []
pred_rows = []
splits_cache = {}

for col in INPUTS:
    for seed in SEEDS:
        tr, va, te = split_grouped_with_val(good.index.values, y_good, amplicon_groups, seed)
        splits_cache[(col, seed)] = (tr, va, te)
        Xtr, Xva, Xte = fit_tfidf(good.iloc[tr][col].values, good.iloc[va][col].values, good.iloc[te][col].values)

        Xtrva = np.vstack([Xtr.toarray(), Xva.toarray()])
        ytrva = np.concatenate([y_good[tr], y_good[va]])
        yte = y_good[te]
        for model_name, clf in [
            ("svm", LinearSVC(class_weight="balanced", random_state=seed, max_iter=20000)),
            ("rf", RandomForestClassifier(n_estimators=400, class_weight="balanced", random_state=seed, n_jobs=-1)),
        ]:
            clf.fit(Xtrva, ytrva)
            pred = clf.predict(Xte)
            rows.append(dict(input=col, seed=seed, model=model_name,
                              f1=f1_score(yte, pred, average="macro"),
                              bal_acc=balanced_accuracy_score(yte, pred)))
            pred_rows.extend([dict(input=col, seed=seed, model=model_name,
                                    test_idx=int(good.index.values[i]), y_true=int(t), y_pred=int(p))
                               for i, t, p in zip(te, yte, pred)])
    pd.DataFrame(rows).to_csv(f"{OUT}/pralone_classical_raw.csv", index=False)
    pd.DataFrame(pred_rows).to_csv(f"{OUT}/pralone_classical_predictions.csv", index=False)
    print(f"[classical] input={col} done")
```

### 6. CNNs, PR alone and RT alone, 3 seeds -- needs GPU. Same architectures/training loop as Fragsweep/Nineseed/Ampmcnemar.

```python


class StochasticDepth(Layer):
    def __init__(self, p=0.85, **kw): super().__init__(**kw); self.p = p
    def call(self, x, training=None):
        if not training: return x
        r = tf.random.uniform([tf.shape(x)[0], 1, 1])
        return x * tf.floor(self.p + r) / self.p
    def get_config(self): return {**super().get_config(), "p": self.p}

def res_block(x, f, stride=1, sp=0.85, dr=0.1):
    sc = x
    x = Conv1D(f, 3, strides=stride, padding="same", use_bias=False)(x)
    x = ReLU()(BatchNormalization()(x)); x = SpatialDropout1D(dr)(x)
    x = BatchNormalization()(Conv1D(f, 3, padding="same", use_bias=False)(x))
    if stride != 1 or sc.shape[-1] != f:
        sc = BatchNormalization()(Conv1D(f, 1, strides=stride, padding="same", use_bias=False)(sc))
    return ReLU()(Add()([StochasticDepth(sp)(x), sc]))

def build_resnet18(input_len, n_classes):
    inp = Input((input_len, 1)); x = inp
    x = ReLU()(BatchNormalization()(Conv1D(32, 7, strides=2, padding="same", use_bias=False)(x)))
    x = MaxPooling1D(3, strides=2, padding="same")(x)
    for f, sp in [(32, .95), (32, .92)]:  x = res_block(x, f, 1, sp)
    for f, sp in [(64, .88), (64, .85)]:  x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    for f, sp in [(128, .80), (128, .78)]: x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    for f, sp in [(256, .75), (256, .72)]: x = res_block(x, f, 2 if f != x.shape[-1] else 1, sp)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

def build_vgg16(input_len, n_classes):
    def conv_block(x, f, n_convs):
        for _ in range(n_convs):
            x = ReLU()(BatchNormalization()(Conv1D(f, 3, padding="same", use_bias=False)(x)))
        return MaxPooling1D(2, padding="same")(x)
    inp = Input((input_len, 1))
    x = conv_block(inp, 32, 2)
    x = conv_block(x,   64, 2)
    x = conv_block(x,  128, 3)
    x = conv_block(x,  256, 3)
    x = conv_block(x,  256, 3)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    x = Dropout(0.4)(ReLU()(Dense(256)(x)))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

def inception_module(x, f1, f3r, f3, f5r, f5, fp):
    b1 = Conv1D(f1, 1, padding="same", activation="relu")(x)
    b3 = Conv1D(f3, 3, padding="same", activation="relu")(Conv1D(f3r, 1, padding="same", activation="relu")(x))
    b5 = Conv1D(f5, 5, padding="same", activation="relu")(Conv1D(f5r, 1, padding="same", activation="relu")(x))
    bp = Conv1D(fp, 1, padding="same", activation="relu")(MaxPooling1D(3, strides=1, padding="same")(x))
    return Concatenate()([b1, b3, b5, bp])

def build_inception(input_len, n_classes):
    inp = Input((input_len, 1))
    x = MaxPooling1D(3, strides=2, padding="same")(
            ReLU()(BatchNormalization()(Conv1D(32, 7, strides=2, padding="same")(inp))))
    x = inception_module(x, 16, 24, 32, 4,  8,  8)
    x = inception_module(x, 32, 32, 48, 8, 24, 16)
    x = MaxPooling1D(3, strides=2, padding="same")(x)
    x = inception_module(x, 48, 24, 52, 8, 24, 16)
    x = Dropout(0.4)(GlobalAveragePooling1D()(x))
    out = Dense(n_classes, activation="softmax")(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(3e-4), loss=CategoricalCrossentropy(label_smoothing=0.1), metrics=["accuracy"])
    return m

BUILDERS = {"resnet18": build_resnet18, "vgg16": build_vgg16, "inception": build_inception}
DEEP_MODELS = ["resnet18", "vgg16", "inception"]

def run_deep_original(model_name, Xtr, ytr, Xva, yva, Xte, yte, seed, n_classes):
    """Exact training recipe from the original per-architecture notebooks (verified byte-identical
    in Phase A, same code already used in Fragsweep/Ambig/Enc): per-feature StandardScaler (fit on train
    only), one-hot labels, CategoricalCrossentropy(label_smoothing=0.1), Adam(3e-4), up to 120
    epochs, EarlyStopping(patience=15) + ReduceLROnPlateau(factor=0.5, patience=6, min_lr=1e-6)."""
    tf.random.set_seed(seed)
    sc = StandardScaler()
    Xtr_d = sc.fit_transform(np.asarray(Xtr.todense() if hasattr(Xtr, "todense") else Xtr))[..., None]
    Xva_d = sc.transform(np.asarray(Xva.todense() if hasattr(Xva, "todense") else Xva))[..., None]
    Xte_d = sc.transform(np.asarray(Xte.todense() if hasattr(Xte, "todense") else Xte))[..., None]
    cw = dict(enumerate(compute_class_weight("balanced", classes=np.unique(ytr), y=ytr)))
    model = BUILDERS[model_name](Xtr_d.shape[1], n_classes)
    model.fit(Xtr_d, to_categorical(ytr, n_classes),
              validation_data=(Xva_d, to_categorical(yva, n_classes)),
              epochs=EPOCHS, batch_size=BATCH, class_weight=cw, shuffle=True, verbose=0,
              callbacks=[EarlyStopping("val_loss", patience=15, restore_best_weights=True),
                         ReduceLROnPlateau("val_loss", factor=0.5, patience=6, min_lr=1e-6)])
    prob = model.predict(Xte_d, verbose=0)
    pred = prob.argmax(1)
    f1 = f1_score(yte, pred, average="macro"); bal = balanced_accuracy_score(yte, pred)
    tf.keras.backend.clear_session()
    return f1, bal, pred

DEEP_RAW_PATH = f"{OUT}/pralone_deep_raw.csv"
if os.path.exists(DEEP_RAW_PATH):
    _existing_deep = pd.read_csv(DEEP_RAW_PATH)
    deep_done_keys = set(zip(_existing_deep["input"], _existing_deep["seed"], _existing_deep["model"]))
    deep_rows = _existing_deep.to_dict("records")
    print(f"Resuming deep models: {len(deep_rows)} rows already done.")
else:
    deep_rows = []
    deep_done_keys = set()

deep_pred_rows = []
for (col, seed), (tr, va, te) in splits_cache.items():
    remaining = [m for m in DEEP_MODELS if (col, seed, m) not in deep_done_keys]
    if not remaining:
        continue
    Xtr, Xva, Xte = fit_tfidf(good.iloc[tr][col].values, good.iloc[va][col].values, good.iloc[te][col].values)
    ytr, yva, yte = y_good[tr], y_good[va], y_good[te]
    for m in remaining:
        f1, bal, pred = run_deep_original(m, Xtr, ytr, Xva, yva, Xte, yte, seed, n_cls)
        deep_rows.append(dict(input=col, seed=seed, model=m, f1=f1, bal_acc=bal))
        deep_pred_rows.extend([dict(input=col, seed=seed, model=m,
                                     test_idx=int(good.index.values[i]), y_true=int(t), y_pred=int(p))
                                for i, t, p in zip(te, yte, pred)])
        pd.DataFrame(deep_rows).to_csv(DEEP_RAW_PATH, index=False)   # incremental save
        print(f"[deep] input={col} seed={seed} model={m} f1={f1:.4f} done")
    pd.DataFrame(deep_pred_rows).to_csv(f"{OUT}/pralone_deep_predictions.csv", index=False)

pralone_all = pd.concat([pd.DataFrame(rows), pd.DataFrame(deep_rows)], ignore_index=True)
pralone_all_predictions = pd.concat([pd.DataFrame(pred_rows), pd.DataFrame(deep_pred_rows)], ignore_index=True)
pralone_all.to_csv(f"{OUT}/pralone_all_raw.csv", index=False)
pralone_all_predictions.to_csv(f"{OUT}/pralone_all_predictions.csv", index=False)
print(pralone_all.groupby(["input","model"])[["f1","bal_acc"]].agg(["mean","std"]).round(4))
```
